## 2. Análise de Qualidade dos Dados

```python
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100
pd.DataFrame({'Missing': missing, 'Percent (%)': missing_percent})
```

```python
missing_percent.plot(kind='bar', figsize=(12,6))
plt.title('Porcentagem de valores faltantes por coluna')
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns
plt.figure(figsize=(12,6))
df[numeric_cols].boxplot(rot=45)
plt.title("Boxplots das variáveis numéricas")
plt.show()
```

As colunas com maior probabilidade de outliers são:

likes_count

shares_count

comments_count

channel_subscribers

total_views

Isso é esperado em dados de YouTube: vídeos virais geram distribuições muito assimétricas.

## 3. Análise Univariada

```python
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribuição de {col}')
    plt.show()
```

```python
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    plt.figure(figsize=(6,4))
    df[col].value_counts().plot(kind='bar')
    plt.title(f'Distribuição de {col}')
    plt.show()
```

**Distribuição**

A distribuição é extremamente assimétrica para a direita (right-skewed).

Poucos vídeos possuem muitas visualizações, enquanto a maioria tem visualizações baixas ou moderadas.

Interpretação

Esse comportamento é típico de plataformas como YouTube, onde poucos vídeos viralizam.

## 4. Análise Bivariada

```python
# Selecionar apenas as variáveis numéricas
numeric_df = df.select_dtypes(include=['int64', 'float64'])

# Calcular a matriz de correlação
correlation_matrix = numeric_df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(14,10))
sns.heatmap(correlation_matrix, cmap='viridis', annot=False)
plt.title("Matriz de Correlação entre Variáveis Numéricas")
plt.show()
```

```python
# Selecionar correlações absolutas com a variável alvo
corr_target = correlation_matrix['total_views'].abs().sort_values(ascending=False)
corr_target
```

```python
# Pegar as top 5 variáveis mais correlacionadas (ignorando a própria total_views)
top_numeric_corr = corr_target.index[1:6]

for col in top_numeric_corr:
    plt.figure(figsize=(6,4))
    sns.scatterplot(x=df[col], y=df['total_views'])
    plt.title(f"Relação entre total_views e {col}")
    plt.xlabel(col)
    plt.ylabel("total_views")
    plt.show()
```

```python
# Selecionar colunas categóricas
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    plt.figure(figsize=(8,5))
    sns.boxplot(x=df[col], y=df['total_views'])
    plt.title(f"Distribuição de total_views por categoria de {col}")
    plt.xticks(rotation=45)
    plt.xlabel(col)
    plt.ylabel("total_views")
    plt.show()
```
